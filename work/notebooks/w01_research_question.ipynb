```python
df.columns.tolist()
```

```python
#1. Stale Pages
stale_pages = (df["days_since_last_update"] >= 180).sum()
print(stale_pages)
#Explanation 1: 174 pages that have not been updated for the past 180 days

#2. Older comand with meaningful search demand
demand = ((df["days_since_last_update"] >= 90) & (df["impressions_90d"] >=  500)).sum()
print(demand)
#Explanation 2: 6575 pages have not been updated for at least 90 days and still receive more than 500 impressions

#3. Lower Recent Clicks
df["expected_clicks_30d"] = df["clicks_90d"] / 3
df["lower_clicks"] = (df["clicks_prev_30d"] < df["expected_clicks_30d"])
lower_click = df["lower_clicks"].sum()
print(lower_click)

#Explanation 3: 9701 pages had previous 30 day clicks below one-third of their 90 day click total.
```
