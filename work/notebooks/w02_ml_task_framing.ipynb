## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
'''
A fixed rule can work for simple situations, but this project may involve many interacting signals such as impressions, freshness, CTR, ranking positions and recent performance. Writing enough hand-crafted if-statements for big data can become extremely difficult to maintain and some patterns may be missed.

For reference, traditional programming uses predefined rules with data to produce an answer (Rules + Data -> Answer). whileas machine learning learns patterns from examples in the data and uses those patterns to make good predictions or decisions (Data + Example -> Learned Pattern -> Predictions / Decisions)

If a learned model is able to combine signals such as high impressions, moderate decline, unusual CTR and older content to produce a better ranked review queue than a transparent rule-based baseline, then additional complexity of ML is justified.
'''
```
